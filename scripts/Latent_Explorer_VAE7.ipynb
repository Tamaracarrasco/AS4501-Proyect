# Explorador interactivo del espacio latente (vae7)

Prototipo: proyeccion UMAP 2D del latente de `vae7`
(`file_out_data/checkpoints/vae_best7.pt`) donde, al pasar el mouse sobre un
punto, se muestra:
- el stamp original del objeto, para una **banda** (g/r/i/z) y **nivel**
  (0..4) elegibles con dos selectores en vivo, en el propio grafico
- su reconstruccion por el VAE, con la misma banda/nivel
- `desi_id`, redshift espectroscopico y tipo morfologico

Corre inline en el notebook (Bokeh + `output_notebook()`) y tambien se puede
exportar a un `.html` standalone (todo embebido con `INLINE` resources, sin
dependencias externas, selectores incluidos) para compartir sin necesitar un
kernel vivo.

**Como se logra que banda/nivel sean interactivos sin recalcular nada en
Python**: cada objeto se codifica una sola vez como un **atlas** — una unica
imagen en mosaico con las 4 bandas x 5 niveles (20 tiles). El selector de
banda/nivel corre en el navegador (`CustomJS`) y solo recorta visualmente el
tile activo (`background-position` en CSS). Asi el peso sigue siendo ~2
imagenes por punto (input + reconstruccion) en vez de 20x2.

**Sobre el tamano (leer antes de subir `N_POINTS`)**: aun con el truco del
atlas, para **todo** el dataset (21285 objetos x 2 atlas) el archivo
resultante seria pesado — no escala bien como HTML autocontenido. Este
prototipo arranca con un subconjunto chico (`N_POINTS`) para que veas el
tamano real primero; al final hay notas sobre como escalar.

## 3. Cargar vae7, codificar el latente y proyectar en 2D

```python
import numpy as np
import torch

from VAE import MultiResVAE, _to_display

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"device: {device}")

npz = np.load(npz_path, allow_pickle=True)
X_all = npz["X"]                                    # (N, 5, 4, 30, 30)
ids_all = np.array([str(i) for i in npz["ids"]])
type_all = np.asarray(npz["type"]).astype(str)
z_all = np.asarray(npz["z"], dtype=np.float32)
idx_train = np.asarray(npz["idx_train"])
idx_val = np.asarray(npz["idx_val"])
idx_test = np.asarray(npz["idx_test"])

split_all = np.array(["train"] * len(ids_all))
split_all[idx_val] = "val"
split_all[idx_test] = "test"

ckpt_path = repo_root / "file_out_data" / "checkpoints" / "vae_best7.pt"
if not ckpt_path.exists():
    raise FileNotFoundError(f"No se encontro el checkpoint: {ckpt_path}")

ckpt = torch.load(ckpt_path, map_location=device)
config = dict(ckpt["config"])
model = MultiResVAE(config).to(device)
model.load_state_dict(ckpt["model"])
model.eval()

print(f"vae7: z_dim={config['z_dim']} | N={len(ids_all)}")
```

```python
@torch.no_grad()
def encode_all(model, X, device, batch_size=256):
    mus = []
    for i in range(0, len(X), batch_size):
        xb = torch.as_tensor(X[i:i + batch_size], dtype=torch.float32).to(device)
        mu, _ = model.encode(xb)
        mus.append(mu.cpu().numpy())
    return np.concatenate(mus, axis=0)


mu_all = encode_all(model, X_all, device)

try:
    import umap
    emb = umap.UMAP(n_components=2, random_state=42).fit_transform(mu_all)
    proj_method = "UMAP"
except Exception:
    from sklearn.decomposition import PCA
    emb = PCA(n_components=2).fit_transform(mu_all)
    proj_method = "PCA (umap no instalado)"

print(f"proyeccion 2D: {proj_method} | mu.shape={mu_all.shape} | emb.shape={emb.shape}")
```

## 4. Parametros: subconjunto y tamano del atlas

Banda (g/r/i/z) y nivel (0..4) ahora son **interactivos**: se eligen con dos
selectores en el propio grafico (seccion 6), sin volver a correr celdas.
`DEFAULT_BAND`/`DEFAULT_LEVEL` solo fijan que se muestra al abrir el grafico.

Empieza chico (`N_POINTS` bajo) para ver el tamano real del HTML antes de
subirlo.


```python
# --- parametros a ajustar ---
SUBSET = "all"           # "all", "train", "val", "test"
N_POINTS = 400            # tope de puntos con atlas embebido (subsample si hay mas)
SCALE = 3                  # zoom del tile en el hover (nativo 30px -> 30*SCALE en pantalla)
DEFAULT_BAND = "r"         # banda mostrada al abrir el grafico (cambiable despues con el selector)
DEFAULT_LEVEL = 2          # nivel mostrado al abrir (0=mayor resolucion .. 4=menor resolucion)
SEED = 42

if SUBSET == "all":
    subset_idx = np.arange(len(ids_all))
else:
    subset_idx = {"train": idx_train, "val": idx_val, "test": idx_test}[SUBSET]

rng = np.random.default_rng(SEED)
if len(subset_idx) > N_POINTS:
    subset_idx = rng.choice(subset_idx, size=N_POINTS, replace=False)
subset_idx = np.sort(subset_idx)

print(f"subset: {SUBSET} | puntos con atlas: {len(subset_idx)}/{len(ids_all)}")
```

## 5. Atlas banda x nivel (input + reconstruccion) en base64

```python
import base64
import io

from PIL import Image

from VAE import BAND_NAMES, N_LEVELS

TILE_NATIVE = 30                          # tamano nativo del stamp (30x30)
ATLAS_W = TILE_NATIVE * len(BAND_NAMES)   # columnas = bandas (g,r,i,z)
ATLAS_H = TILE_NATIVE * N_LEVELS          # filas = niveles (0..4)
TILE_PX = TILE_NATIVE * SCALE             # tamano del tile ya escalado, en pantalla


@torch.no_grad()
def reconstruct(model, X, device, batch_size=256):
    recons = []
    for i in range(0, len(X), batch_size):
        xb = torch.as_tensor(X[i:i + batch_size], dtype=torch.float32).to(device)
        r, _, _ = model(xb)
        recons.append(r.cpu().numpy())
    return np.concatenate(recons, axis=0)


def build_atlas_uri(X_obj):
    """X_obj: (5,4,30,30) de UN objeto -> atlas PNG (ATLAS_W x ATLAS_H) con
    las 4 bandas (columnas) x 5 niveles (filas). Cada tile se estira con el
    mismo criterio que scripts/VAE.py (_to_display, modo 'band': percentil
    1-99 POR TILE), para consistencia visual con el resto del proyecto."""
    atlas = Image.new("L", (ATLAS_W, ATLAS_H))
    for level in range(N_LEVELS):
        for band_idx in range(len(BAND_NAMES)):
            disp = _to_display(X_obj[level], "band", band_idx)     # float [0,1], HxW
            tile = np.clip(disp * 255, 0, 255).astype(np.uint8)
            atlas.paste(Image.fromarray(tile), (band_idx * TILE_NATIVE, level * TILE_NATIVE))
    buf = io.BytesIO()
    atlas.save(buf, format="PNG", optimize=True)
    return "data:image/png;base64," + base64.b64encode(buf.getvalue()).decode("ascii")


X_subset = X_all[subset_idx]
recon_subset = reconstruct(model, X_subset, device)

atlas_input = [build_atlas_uri(X_subset[i]) for i in range(len(subset_idx))]
atlas_recon = [build_atlas_uri(recon_subset[i]) for i in range(len(subset_idx))]

avg_kb = np.mean([len(u) for u in atlas_input + atlas_recon]) / 1e3 * 0.75  # base64 -> bytes aprox
print(f"atlas generados: {len(atlas_input)} input + {len(atlas_recon)} recon "
      f"(~{avg_kb:.1f} KB c/u en promedio, {ATLAS_W}x{ATLAS_H}px, "
      f"{len(BAND_NAMES)} bandas x {N_LEVELS} niveles)")
```

## 6. Grafico interactivo (Bokeh): selectores de banda y nivel + hover

```python
from bokeh.io import output_notebook, show
from bokeh.layouts import column, row
from bokeh.models import CategoricalColorMapper, ColumnDataSource, CustomJS, HoverTool, Select
from bokeh.palettes import Category10
from bokeh.plotting import figure

output_notebook()

default_band_idx = BAND_NAMES.index(DEFAULT_BAND)
default_pos = f"-{default_band_idx * TILE_PX}px -{DEFAULT_LEVEL * TILE_PX}px"

types_unique = sorted(np.unique(type_all[subset_idx]).tolist())
base_palette = Category10[10]
palette = [base_palette[i % 10] for i in range(len(types_unique))]
color_map = CategoricalColorMapper(factors=types_unique, palette=palette)

source = ColumnDataSource(data=dict(
    x=emb[subset_idx, 0], y=emb[subset_idx, 1],
    desi_id=ids_all[subset_idx], z=z_all[subset_idx].astype(float),
    type=type_all[subset_idx],
    atlas_input=atlas_input, atlas_recon=atlas_recon,
    bg_pos_input=[default_pos] * len(subset_idx),
    bg_pos_recon=[default_pos] * len(subset_idx),
))

# background-size fija a la escala del atlas completo; background-position
# (que cambia el CustomJS de abajo) recorta solo el tile activo.
bg_size = f"{ATLAS_W * SCALE}px {ATLAS_H * SCALE}px"
tile_css = (
    f"width:{TILE_PX}px;height:{TILE_PX}px;overflow:hidden;"
    f"background-repeat:no-repeat;background-size:{bg_size};image-rendering:pixelated;"
)
hover_html = f"""
<div style="max-width:260px">
    <div style="display:flex; gap:6px;">
        <div style="text-align:center;">
            <div style="{tile_css}background-image:url(@atlas_input);background-position:@bg_pos_input;"></div>
            <div style="font-size:9px;">input</div>
        </div>
        <div style="text-align:center;">
            <div style="{tile_css}background-image:url(@atlas_recon);background-position:@bg_pos_recon;"></div>
            <div style="font-size:9px;">recon</div>
        </div>
    </div>
    <div style="margin-top:4px;"><b>id:</b> @desi_id</div>
    <div><b>z_spec:</b> @z{{0.0000}}</div>
    <div><b>tipo:</b> @type</div>
</div>
"""

band_select = Select(title="Banda", value=DEFAULT_BAND, options=BAND_NAMES)
level_select = Select(title="Nivel (0 = mayor resolucion, 4 = menor resolucion)",
                       value=str(DEFAULT_LEVEL), options=[str(i) for i in range(N_LEVELS)])

# Corre en el navegador (no en Python): al cambiar banda/nivel, recalcula el
# offset del tile activo dentro del atlas y lo escribe en la MISMA fila para
# todos los puntos (la seleccion es global, no por punto). No hace falta
# volver a tocar el kernel, y funciona igual en el .html exportado.
band_level_callback = CustomJS(
    args=dict(source=source, band_select=band_select, level_select=level_select,
              tile_px=TILE_PX, bands=BAND_NAMES),
    code="""
        const band_idx = bands.indexOf(band_select.value);
        const level_idx = parseInt(level_select.value);
        const pos = "-" + (band_idx * tile_px) + "px -" + (level_idx * tile_px) + "px";
        const n = source.data['x'].length;
        source.data['bg_pos_input'] = new Array(n).fill(pos);
        source.data['bg_pos_recon'] = new Array(n).fill(pos);
        source.change.emit();
    """,
)
band_select.js_on_change("value", band_level_callback)
level_select.js_on_change("value", band_level_callback)

p = figure(width=850, height=650, tools="pan,wheel_zoom,reset,box_zoom",
           title=f"vae7 - espacio latente ({proj_method})")
p.add_tools(HoverTool(tooltips=hover_html))
p.scatter("x", "y", source=source, size=8, alpha=0.75, marker="circle",
          color={"field": "type", "transform": color_map}, legend_field="type")
p.legend.location = "top_left"
p.legend.click_policy = "hide"
p.xaxis.axis_label = f"{proj_method} 1"
p.yaxis.axis_label = f"{proj_method} 2"

layout = column(row(band_select, level_select), p)
show(layout)
```

## Notas / como escalar

- El atlas empaqueta 20 tiles (4 bandas x 5 niveles) por objeto en **una**
  imagen; el selector solo recorta el tile activo con CSS
  (`background-position`), corriendo en el navegador via `CustomJS` — por
  eso banda/nivel son interactivos sin volver a tocar Python ni perder la
  exportacion a `.html` autonomo.
- Si aun asi `N_POINTS` grande hace el HTML demasiado pesado para compartir,
  las alternativas para llegar a **todo** el dataset (21285 objetos) son, de
  mas simple a mas trabajo:
  1. Mostrar un solo split (`test`, ~2129 objetos) en vez de `all`, o bajar
     `SCALE` — mismo notebook, solo cambiar los parametros de la seccion 4.
  2. Mostrar solo el atlas `input` en el hover (sin `recon`) para
     aproximadamente la mitad del peso.
  3. Pasar a atlas generados **bajo demanda**: un Bokeh server (o un widget
     de Jupyter con `ipywidgets` + callback en Python que llame a
     `build_atlas_uri` solo para el punto bajo el cursor) en vez de
     embeber los de los 21285 objetos de antemano. Escala sin limite de
     tamano, pero deja de ser un `.html` 100% autonomo: necesita un
     kernel/servidor Python vivo para responder al hover.
- El stretch por tile (percentil 1-99) es el mismo que usa
  `scripts/VAE.py` (`_to_display`, modo `"band"`), para que las imagenes se
  vean igual que en las grillas de reconstruccion que ya genera el
  entrenamiento.
- El color por `type` tiene leyenda clickeable (click para ocultar/mostrar
  una clase). Si en cambio quieres colorear por redshift (continuo), avisa y
  se agrega un segundo panel como el de `latent_umap()` en
  `scripts/SemiSupervVAE.py`.
