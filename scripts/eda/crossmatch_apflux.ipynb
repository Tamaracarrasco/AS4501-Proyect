### En este script, buscaré hacer crossmatch del dataframe que contiene features de los stamps
con otro dataframe que contiene las columnas de los flujos para aperturas de distinto radio (Obtenidos del tractor catalog)

- los radios de apertura que se consideran son hasta un radio de 3.5 [arcsec], esto es porque s eprobarán en el nivel con mejor resolución.
- solo se agregan los flujos de apertura considerando banda g y r.

![imagen de apertura de flujos](../../figures/aperture_fluxes.jpg)


```python
## Importación de librerías

import numpy as np
import pandas as pd
```

```python
df1 = pd.read_csv("../../data/crossmatch_apflux.txt", sep=",")
df1.info()
```

```python
df1.columns.tolist()
```

```python
df2 = pd.read_csv("../../data/features_images_20260618.csv", sep=",")
df2.info()
```

```python
df1.rename(columns={"id_1": "id"}, inplace=True)
```

```python
cols = [
'apflux_g_0_5',
 'apflux_g_0_75',
 'apflux_g_1_0',
 'apflux_g_1_5',
 'apflux_g_2_0',
 'apflux_g_3_5',
 'apflux_r_0_5',
 'apflux_r_0_75',
 'apflux_r_1_0',
 'apflux_r_1_5',
 'apflux_r_2_0',
 'apflux_r_3_5',
]

df2 = df2.merge(df1[["id"] + cols], on="id", how="left")
```
